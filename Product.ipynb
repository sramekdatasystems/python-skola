```python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

products = [Product("Kování", 25, 10),
            Product("Nůžky", 90, 5),
            Product("Nůž", 250, 3)]

for product in products:
    total_price = product.price * product.quantity
    print(f"{product.name} celkově stojí {total_price}")




```
