```python
import math

number_1 = int(input("Zadejte dělence:\n"))
number_2 = int(input("Zadejte dělitele:\n"))

number = number_1 / number_2

number_1 = math.floor(number)
number_2 = number - number_1

print(f"výsledek je celé číslo {number_1} a zbylé desetiny {number_2}")
```
