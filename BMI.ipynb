```python
#BMI calculation and status assessment
weight = float(input("Zadejte svou hmotnost v Kg: "))
height = float(input("Zadejte svou výšku v metrech: "))
bmi = round(weight / (height * height), 1)

if bmi < 18.5:
    print(f"Vaše BMI je {bmi}. To je podváha")
elif bmi < 25:
    print(f"Vaše BMI je {bmi}. To je normální váha")
elif bmi < 30:
    print(f"Vaše BMI je {bmi}. To je nadváha")
else:
    print(f"Vaše BMI je {bmi}. To je obezita")


```
