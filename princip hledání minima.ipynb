```python
# program to display the lowest set temperature

# input directly into the list
temperatures_list = input("zadejte teploty oddělené mezerou: ").split(", ")
temperatures_list_number = []


# cycle to set the string in the list to a number
for index in temperatures_list:
    temperatures_list_number.append(int(index))

minimum = temperatures_list_number[0]
for one_number in temperatures_list_number:
    if one_number < minimum:
        minimum = one_number

print(f"Nejmenší zadaná teplota je {minimum} C")
```
