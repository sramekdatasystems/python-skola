```python
# rock_scissors_paper
import random

# rock paper scissors choice
user = int(input("Zadejte pro kámen (0) , pro nůžky (1) a pro papír (2):\n"))
computer = random.randint(0, 2)

# loading symbols into variables
rock = '''
    '   ____)
      (_____)
      (_____)
      (____)
---.__(___)
'''

paper = '''
    _______
---'     ____)
          ______)
          _______)
         _______)
  ---.__________)
'''

scissors = '''
    _______
  ---' ____)____
          ______)
       __________)
      (____)
 --.__(___)
'''

# listing selection and symbol user
if user == 0:
    print("Tvůj výběr je kámen")
    print(rock)
elif user == 1:
    print("Tvůj výběr jsou nůžky")
    print(scissors)
elif user == 2:
    print("Tvůj výběr je papír")
    print(paper)

# listing selection and computer symbol
if computer == 0:
    print("Výběr počítače je kámen")
    print(rock)
elif computer == 1:
    print("Výběr počítače jsou nůžky")
    print(scissors)
elif computer == 2:
    print("Výběr počítače je papír")
    print(paper)


# result evaluation
if (user == 0 and computer == 1) or (user == 1 and computer == 2) or (user == 2 and computer == 0):
    print("Vyhrál jste a počítač prohrál")
elif (user == 0 and computer == 2) or (user == 1 and computer == 0) or (user == 2 and computer == 1):
    print("Prohrál jsi a počítač vyhrál")
else:
    print("Remíza")
```
