```python
# after entering sales, the program will list the smallest and largest entered sales

# entry in the list
sales_list = input("Zadejte tržby oddělené čárkou a mezerou:\n").split(", ")
sales_list_number = []

# converting list contents to numbers
for sales in sales_list:
    sales_list_number.append(int(sales))

# calculating the minimum using the day counter
minimum = sales_list_number[0]
counter = 0
day_min = 1
for sales in sales_list_number:
    counter += 1
    if sales < minimum:
        minimum = sales
        day_min = counter

# calculating the maximum using the day counter
maximum = sales_list_number[0]
counter = 0
day_max = 1
for sales in sales_list_number:
    counter += 1
    if sales > maximum:
        maximum = sales
        day_max = counter

 # listing the worst and best days and their order
print(f"Nejhorší den byl {day_min}. den s tržbou {minimum} Kč\n a nejlepší byl {day_max}. den s tržbou {maximum} Kč")
```
