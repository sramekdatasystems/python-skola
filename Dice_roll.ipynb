```python
# circle content
import math
print("Zadejte poloměr kruhu a program Vám spočítá jeho obsah")
# radius input from the user
radius_circle = float(input("Zadejte poloměr kruhu:\n"))

# content calculation
circle_content = math.pi * (radius_circle ** 2)

print(f"Obsah kruhu s rádiusem {radius_circle} je {circle_content:.2f}")

```
