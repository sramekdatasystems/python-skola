```python
#modulo3
total_minutes = int(input("Zadejte počet minut:\n"))

hours = total_minutes // 60
minutes = total_minutes % 60

if minutes == 0:
    print(f"zadaný počet minut {total_minutes} je {hours} h")
else:
    print(f"zadaný počet minut {total_minutes} je {hours} h a {minutes} m")
```
