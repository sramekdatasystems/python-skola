```python
# average_value_of_taxable_amounts

height = input("Zadejte průměrné výšky oddělené čárkou a mezerou:\n")
list_height = height.split(", ")
list_height_number = list(map(int, list_height))

# resetting counters
counter = 0
sum_heights = 0

# sum of heights
for one_height in list_height_number:

    sum_heights = sum_heights + one_height
    counter += 1

# calculation
average_height = sum_heights / counter

# result display
print(f"Průměrná výška je: {average_height:.1f}")



```
