```python
# cvičení s While 2

print("Zadávejte celá čísla a až přesáhne součet zadaných čísel 100, tak program vypíše počet zadaných čísel.")

total = 0
counter = 0
while total < 100:
    number = int(input("Zadejte číslo:\n"))
    total += number
    counter += 1

print(f"Součet Vámi zadaných čísel je {total} a je to součet {counter} čísel")

```
