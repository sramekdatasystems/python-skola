```python
# coin_toss_percentage
import random

# random decimal place assignment
coin_toss = random.random()

# up to and including 0.5 it is Heads and above 0.5 it is Tails
if coin_toss <= 0.5:
    print("Panna")
else:
    print("Orel")

```
