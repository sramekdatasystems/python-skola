```python
!pip install unidecode -q

from unidecode import unidecode


my_distr = {
    "zbynek" : "602564048",
    "pavlina" : "602174045"
}
while True:


    name = input("Čí číslo chcete hledat?,\n nebo chcete konec(zadejte konec)\n")
    name_oci = unidecode(name).lower()

    if name_oci == "konec":
        print("Konec programu")
        break

    for key in my_distr:
        if key == name_oci:
            print(my_distr[key])

```
