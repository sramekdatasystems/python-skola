```python
#pizza order
pizza_size = input("Zadejte jakou velikost pizzy chcete(S/M/L):\n").lower()

#valid selection testing
if pizza_size == "s" or pizza_size == "m" or pizza_size == "l":
    #size specification
    if pizza_size == "s":
        bill = int(100)
    elif pizza_size == "m":
        bill = int(150)
    else:
        bill = int(200)

    hot_peppers = input("Chcete na pizzu feferonky? (Ano/Ne)\n").lower()
    #valid selection testing
    if hot_peppers == "ano" or hot_peppers == "ne":
        #selection of hot peppers
        if hot_peppers == "ano":
            if pizza_size == "s":
                bill += 20
            else:
                bill += 30
    else:
        print(f"zadali jste {hot_peppers}, to není platný výběr")

    extra_cheese = input("Chcete sýr navíc? (Ano/Ne)\n").lower()
    #valid selection testing
    if extra_cheese == "ano" or extra_cheese == "ne":
        if extra_cheese == "ano":
            bill += 15

    else:
        print(f"zadali jste {extra_cheese}, to není platný výběr")
    print(f"celková cena pizzy je {bill} Kč")
else:
    print(f"zadali jste {pizza_size}, to není platný výběr")
```
