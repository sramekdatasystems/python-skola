```python
# Caesarova šifra

# seznam všech malých písmen abecedy, slouží k nalezení pozice (indexu) písmene v abecedě
alphabet = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']

# funkce, která podle hodnoty "action" buď zakóduje, nebo dekóduje zprávu
def coding(action):
    # opakuj, dokud uživatel nezadá platné celé číslo
    while True:
        try:
            # pokus o načtení vstupu a jeho převod na celé číslo
            word_shift = int(input("Zadejte posun:\n"))

            # pokud převod proběhne úspěšně, cyklus se ukončí
            break

        except ValueError:
            # pokud uživatel nezadá číslo, vyvolá se chyba ValueError a cyklus pokračuje
            print("To nebylo celé číslo! Zkuste to znovu.")

    # dekódování je vlastně jen kódování s opačným (záporným) posunem
    if action == "decoding":
        word_shift *= -1

    # sem se bude postupně skládat výsledný text, písmeno po písmenu
    coded_word = ""
    # načtení zprávy ke zpracování, převedené na malá písmena, aby vždy sedělo hledání v abecedě
    message = input("Zadejte zprávu ke kódování:\n").lower()

    # projdi zprávu znak po znaku
    for letter in message:
        # pokud je znak mezera, rovnou ji zkopírujeme beze změny
        if letter == " ":
            coded_word += " "
            continue  # přeskoč zbytek tohoto průchodu cyklem, jdi na další písmeno

        # hledání pozice tohoto písmene v abecedě
        for index in range(len(alphabet)):
            if letter == alphabet[index]:
                # posun pozice a "zacyklení" pomocí modula, aby se index nedostal mimo abecedu
                new_index = (index + word_shift) % len(alphabet)
                # přidání posunutého písmene do výsledku
                coded_word += alphabet[new_index]
                break  # písmeno nalezeno a zpracováno, není třeba hledat dál

    # vypiš jinou hlášku podle toho, jestli šlo o kódování, nebo dekódování
    if action == "decoding":
        print(f"Dekódovaná zpráva je:\n{coded_word}")
    else:
        print(f"Zakódovaná zpráva je:\n{coded_word}")


# hlavní smyčka programu — běží, dokud se uživatel nerozhodne skončit
while True:
    # zeptej se uživatele, jestli chce kódovat, nebo dekódovat
    action = input("Chcete dekodovat(decoding) nebo kódovat(coding) zprávu?\n")

    # opakuj dotaz, dokud vstup není přesně "decoding" nebo "coding"
    while action not in ["decoding", "coding"]:
        action = input("Chcete dekodovat(decoding) nebo kódovat(coding) zprávu?\n")

    # zavolej funkci coding s vybranou akcí
    if action == "decoding":
        coding("decoding")
    else:
        coding("coding")

    # --- dotaz, jestli chce uživatel pokračovat ---
    continue_game = input("Chcete pokračovat znovu? (ano / ne):\n").lower()

    # pokud odpověď není "ano", ukonči hlavní smyčku pomocí break
    if continue_game != "ano":
        print("Děkuji za použití programu. Nashledanou!")
        break
```
