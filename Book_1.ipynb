```python
class Book:

    def __init__(self, title, author):
        self.title = title
        self.author = author

book_1 = Book("Jaro doma", "J.Ander")
print(f"Název knihy je {book_1.title} a jejím autorem je {book_1.author}")

book_2 = Book("Myšák", "K.Heller")
print(f"Název knihy je {book_2.title} a jejím autorem je {book_2.author}")
```
