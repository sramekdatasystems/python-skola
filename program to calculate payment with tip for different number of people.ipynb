```python
#program to calculate payment with tip for different number of people
payment = float(input("Kolik máte celkem zaplatit?\n"))
tip = float(input("Kolik chcete dát spropitného (v %)?\n"))
number_people = int(input("Mezi kolik lidí se má platba rozdělit?\n"))
#calculate total payment with tip, then split it among all people
print(f"Každý člověk by měl zaplatit {((payment *(1 + (tip/100))) / number_people):.2f}")
```
