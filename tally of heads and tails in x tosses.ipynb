```python
# tally of heads and tails in x tosses
import random
number_throws = int(input("Zadejte kolikrát chcete házet mincí:\n"))
heads_count = 0
tails_count = 0

# a loop that will repeat as many times as the user specified
for i in range(number_throws):
    # randomly selects a number to which Heads or Tails is assigned based on a condition
    coin = random.randint(0, 1)
    if coin == 0:
        print("Hodil jsi Pannu")
        heads_count += 1
    else:
        print("Hodil jsi Orla")
        tails_count += 1

print(f"Z {number_throws} hodů Panna padla {heads_count} krát a Orel padl {tails_count} krát")
```
