```python
# prints the highest score of a student from the list and their rank

# entry in the list
score_list = input("Zadejte výsledky studentů oddělené čárkou a mezerou:\n").split(", ")
score_list_number = []

# converting list contents to numbers
for score in score_list:
    score_list_number.append(int(score))

# calculating the maximum using the day counter
maximum = score_list_number[0]
counter = 0
day_max = 1
for score in score_list_number:
    counter += 1
    if score > maximum:
        maximum = score
        day_max = counter

 # the ranking of the student with the best score and his points
print(f"Nejlepší skóre měl {day_max}. student se skórem {maximum} bodů")
```
