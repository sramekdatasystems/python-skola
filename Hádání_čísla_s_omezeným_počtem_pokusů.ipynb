```python
# number guessing with a limited number of attempts
import random

print("Máte sedm pokusů na uhodnutí čísla")

counter = 1
# selection of a random number
random_number = random.randint(1, 50)
# initial user input before the loop
user_number = int(input("Zadejte číslo:\n"))

# the loop runs if attempts remain and the user has not guessed the number yet
while counter < 7 and random_number != user_number:
    # hint whether the guess is too low or too high
    if random_number > user_number:
        print("Málo")
    else:
        print("Moc")
    user_number = int(input("Špatně!!!\nZkuste znovu:\n"))
    # increment the attempt counter
    counter += 1

# exit conditions
if random_number == user_number:
    print(f"Uhádl jsi. Správné číslo bylo: {random_number}")
else:
    print(f"Neuhádl jsi. Správné číslo bylo: {random_number}")
```
