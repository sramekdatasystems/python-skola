```python
#program na trojúhelníky(podmínky)
# program to check and classify a triangle based on side lengths
first_side = int(input("Zadejte první stranu trojúhelníku:\n"))
second_side = int(input("Zadejte druhou stranu trojúhelníku:\n"))
third_side = int(input("Zadejte třetí stranu trojúhelníku:\n"))

# check triangle inequality: sum of any two sides must exceed the third side
if first_side + second_side > third_side and first_side + third_side > second_side and second_side + third_side > first_side:
    if first_side == second_side and second_side == third_side:
        print("Trojúhelník je rovnostranný")
    elif first_side == second_side or second_side == third_side or first_side == third_side:
        print("Trojúhelník je rovnoramenný")
    else:
        print("Trojúhelník je různostranný")
else:
    print("Toto není platný trojúhelník.")
```
