```python
znamka = int(input("Zadejte známku: "))
if znamka == 1:
    print("Výborně")
elif znamka == 2:
    print("Chvalitebně")
elif znamka == 3:
    print("Dobře")
elif znamka == 4:
    print("Dostatečně")
elif znamka == 5:
    print("Nedostatečně")
else:
    print("Neplatná známka")
```
