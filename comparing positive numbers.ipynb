```python
#comparing positive numbers
print("Zadejte tři čísla a program vám napíše zda jsou všechna kladná")
number_1 = int(input("Zedejte první číslo: "))
number_2 = int(input("Zadejte druhé číslo: "))
number_3 = int(input("Zadejte třetí číslo: "))

if number_1 > 0 and number_2 > 0 and number_3 > 0:
    print(f"čísla {number_1}, {number_2} a {number_3} jsou všechna kladná")
else:
    print("Všechna čísla nejsou kladná")


```
