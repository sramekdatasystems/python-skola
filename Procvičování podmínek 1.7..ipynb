```python
#zadá se výška a váha a program spočítá BMI a vypíše stav
weight = float(input("Zadejte svou hmotnost v Kg: "))
height = float(input("Zadejte svou výšku v metrech: "))
BMI = weight / (height * height)

if BMI < 18.5:
    print("Podváha")
elif BMI < 25:
    print("Normální váha")
elif BMI < 30:
    print("Nadváha")
else:
    print("Obezita")
```
