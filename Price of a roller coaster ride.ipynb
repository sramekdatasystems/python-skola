```python
#Price of a roller coaster ride
print("Vítejte na horské dráze")
height = int(input("Jaká je Vaše výška v cm?\n"))

#the condition tests the height
if height >= 87:
    print("Můžete jet na horské dráze")
    age = int(input("Zadejte svůj věk:\n"))

    #condition tests age
    if age >= 18:
        bill = 150
    elif age >= 12:
        bill = 100
    else:
        bill = 50
    photo = input("Chcete se nechat vyfotit? Ano/Ne\n").lower()

    #testing if he wants a photo
    if photo == "ano":
        print(f"Cena Vaší vstupenky je {bill + 50} Kč")
    else:
        print(f"Cena Vaší vstupenky je {bill} Kč")


else:
    print("Omlouváme se, ale na horské dráze jet nemůžete")
```
