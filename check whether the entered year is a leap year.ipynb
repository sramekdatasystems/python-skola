```python
#check whether the entered year is a leap year
year = int(input("Zadejte rok:\n"))

#checking divisibility by four
if year % 4 == 0:
    #check for divisibility of hundreds
    if year % 100 > 0:
        print(f"Rok {year} je přestupný")
    else:
        #checking divisibility by four hundred
        if year % 400 == 0:
            print(f"Rok {year} je přestupný")
        else:
            print(f"Rok {year} není přestupný")


else:
    print(f"Rok {year} není přestupný")



```
