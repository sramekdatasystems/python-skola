```python
# Caesar cipher
alphabet = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']

while True:
    action = input("Chcete dekodovat(decoding) nebo kódovat(coding) zprávu?\n")

    if action == "coding":
        while True:
            try:
                # Pokus o načtení vstupu a jeho převod na celé číslo
                word_shift = int(input("Zadejte posun:\n"))

                # Pokud převod proběhne úspěšně, cyklus se ukončí
                break

            except ValueError:
                # Pokud uživatel nezadá číslo, vyvolá se chyba ValueError a cyklus pokračuje
                print("To nebylo celé číslo! Zkuste to znovu.")

        encoded_word = ""
        message = input("Zadejte zprávu ke kódování:\n").lower()

        for letter in message:
            # Pokud je znak mezera, rovnou ji zkopírujeme a nešifrujeme
            if letter == " ":
                encoded_word += " "
                continue  # Skočí na další písmeno v textu

            for index in range(len(alphabet)):
                if letter == alphabet[index]:
                    # Klíčová oprava: (index + posun) % celková_délka_abecedy
                    # Pokud bude výsledek např. 27 a délka abecedy 26, vrátí to index 1 (písmeno 'b')
                    new_index = (index + word_shift) % len(alphabet)
                    encoded_word += alphabet[new_index]
                    break
        print(f"Zakódovaná zpráva je:\n{encoded_word}")

    elif action == "decoding":
        while True:
            try:
                # Pokus o načtení vstupu a jeho převod na celé číslo
                word_shift = int(input("Zadejte posun:\n"))

                # Pokud převod proběhne úspěšně, cyklus se ukončí
                break

            except ValueError:
                # Pokud uživatel nezadá číslo, vyvolá se chyba ValueError a cyklus pokračuje
                print("To nebylo celé číslo! Zkuste to znovu.")

        decoded_word = ""
        message = input("Zadejte zprávu ke kódování:\n").lower()

        for letter in message:
            # Pokud je znak mezera, rovnou ji zkopírujeme a nešifrujeme
            if letter == " ":
                decoded_word += " "
                continue  # Skočí na další písmeno v textu

            for index in range(len(alphabet)):
                if letter == alphabet[index]:
                    # Klíčová oprava: (index + posun) % celková_délka_abecedy
                    # Pokud bude výsledek např. 27 a délka abecedy 26, vrátí to index 1 (písmeno 'b')
                    new_index = (index - word_shift) % len(alphabet)
                    decoded_word += alphabet[new_index]
                    break
        print(f"Dekódovaná zpráva je:\n{decoded_word}")

    else:
        print("Neplatná volba")
    # --- ZDE SE PTÁME NA POKRAČOVÁNÍ ---
    continue_game = input("Chcete pokračovat znovu? (ano / ne):\n").lower()

    # Pokud uživatel nenapíše 'ano', cyklus while True okamžitě ukončíme pomocí break
    if continue_game != "ano":
        print("Děkuji za použití programu. Nashledanou!")
        break
```
