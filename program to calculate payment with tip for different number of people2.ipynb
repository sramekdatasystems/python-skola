```python
#program to calculate payment with tip for different number of people
payment = float(input("Kolik máte celkem zaplatit?\n"))
if payment <= 0:
    print("Nemůžeš platit záporně nebo nic!")
else:
    tip = float(input("Kolik chcete dát spropitného (v %)?\n"))
    if tip > 100 or tip < 0:
        print("Můžeš zadat jen rozmezí 0-100 %")
    else:
        number_people = int(input("Mezi kolik lidí se má platba rozdělit?\n"))
        if number_people <= 0:
            print("Minimum je jeden člověk")
        else:
            #calculate total payment with tip
            total_with_tip = payment * (1 + (tip / 100))
            #split it among all people
            payment_per_person = total_with_tip / number_people
            print(f"Každý člověk by měl zaplatit {payment_per_person:.2f} Kč")
```
