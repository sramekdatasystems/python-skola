```python
# list
poeple = ["David", "Harry","Ron"]

print(poeple[0])

# měníme položku

poeple[1] = "Hermiona"

print(poeple)

# přidávání

poeple.append("Harry")

print(poeple)

# přidávání více položek

poeple.extend(["Crabe", "Goyle"])

print(poeple)

# odebrání

poeple.remove("Ron")

print(poeple)
```
