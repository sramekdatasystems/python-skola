```python
# hangman

# generátor náhodných čísel
import random

from stages import stages

words_list = ["přemek", "jan", "jana", "pavlína", "zbyněk", "ondřej", "emma"]
random_word = random.choice(words_list)
# print(random_word)

# nahrazení písmen podtržítkem
word_hiden = []
for index in range(0, len(random_word)):
    word_hiden.append("_")

# předělání listu na string
printed_word = ""
for index in word_hiden:
    printed_word += index
print(printed_word)

# nastavení počtu životů
lives = 6
print(stages[lives])

# opakování hry

while "_" in word_hiden:
    letter = input("Zadej písmeno: ").lower()
    for index in range(0, len(random_word)):
        if random_word[index] == letter:
            word_hiden[index] = letter

    # předělání listu na string
    printed_word = ""
    for index in word_hiden:
        printed_word += index
    print(printed_word)

    # kontrola životů
    if letter not in random_word:
        lives -= 1
        print(stages[lives])
        print(f"Zbývá ti {lives} zivotů")
        if lives == 0:
            print("prohrál jsi")
            break

    # kontrola vítězství
    if "_" not in word_hiden:
        print("Vyhrál jsi")






```
