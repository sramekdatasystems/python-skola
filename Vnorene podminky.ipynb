```python
prvnicislo = int(input("Zadejte první číslo: "))
druhecislo = int(input("Zadejte druhé číslo: "))
operace = input("Zadejte operaci: ")

if operace == "+":
    print(f"Součet je: {prvnicislo + druhecislo}")
elif operace == "-":
    print(f"Rozdíl čísel je: {prvnicislo - druhecislo}")
elif operace == "*":
    print(f"Součin čísel je: {prvnicislo * druhecislo}")
elif operace == "/":
    if druhecislo == 0:
        print("nulou nelze dělit")
    else:
        print(f"Podíl čísel je: {prvnicislo / druhecislo}")
else:
    print("Neplatná operace")
```
